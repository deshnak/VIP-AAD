```python
from deap import base, creator, tools
import random
```

```python
creator.create("FitnessMax", base.Fitness, weights=(1.0,))
```

```python
creator.create("Individual", list, fitness=creator.FitnessMax)
```

```python
toolbox = base.Toolbox()
```

```python
toolbox.register("attr_bool", random.randint, 0, 1)
toolbox.attr_bool()
```

```python
toolbox.register("individual", tools.initRepeat, creator.Individual, toolbox.attr_bool, 100)
ind = toolbox.individual()
print(ind)
print(len(ind))
```

```python
toolbox.register("population", tools.initRepeat, list, toolbox.individual)
pop = toolbox.population(2)
print(pop[0])
print(pop[1])
```

```python
def evalOneMax(individual):
    individual.fitness.values = (sum(individual),)
    return individual.fitness.values
print(evalOneMax(pop[0]))
print(evalOneMax(pop[1]))
```

```python
child1, child2 = tools.cxTwoPoint(toolbox.clone(pop[0]), toolbox.clone(pop[1]))
print(child1, evalOneMax(child1))
print(child2, evalOneMax(child2))
```

```python
# This is a very pythonic way of printing out the difference between two lists element by element. 
# It combines several concepts including a zipped set of iterators and a list expansion
print('Differences:', [elem1-elem2 for elem1, elem2 in zip(pop[0], child1)])
# Here zip(list_1, list_2) returns an iterator that steps through both lists together
# elem1, elem2 in zip(list_1, list_2) stores the current value in each list in elem1 and elem2 as it iterates
# [elem1-elem2 for ...] builds a list by subtracting the two values as it goes. 
# The final result is an elementwise difference
differences = []
for elem1, elem2 in zip(pop[0], child1):
    differences.append(elem1-elem2)
```

```python
mutant, = tools.mutFlipBit( toolbox.clone(child1),indpb=0.05)
print(mutant, evalOneMax(mutant))
```
